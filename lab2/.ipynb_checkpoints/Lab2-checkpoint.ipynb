```python
NASA_KEY = os.getenv("NASA_API_KEY")

def get_useragent():
    url = "https://httpbin.org/user-agent"
    r = requests.get(url)
    useragent = json.loads(r.text)['user-agent']
    return useragent
```

```python
params = {"start_date": '2024-10-09', 'api_key':NASA_KEY}
headers = {"From": "wfl9zy@virginia.edu", "User-Agent":get_useragent()}
data = requests.get("https://api.nasa.gov/neo/rest/v1/feed", params=params).json()
```

```python
df_neo_dates = pd.json_normalize(data['near_earth_objects']).columns
```

```python
neo_dates = pd.json_normalize(data['near_earth_objects']).columns

df_neo = pd.DataFrame()
for date in neo_dates:
    df_neo = pd.concat([df_neo, pd.json_normalize(data['near_earth_objects'][date])], axis=1)
```

```python
# The general / endpoint on Genius.com is not disallowed in https://genius.com/robots.txt so it should be ok for us to webscrape this
```

```python
# returns cleaned up lyrics from a Genius.com endpoint
def get_clean_lyrics(url):

    response = requests.get(url)
    soup1 = BeautifulSoup(response.text)

    lyrics_div = str(soup1.find('div', class_='Lyrics__Container-sc-1ynbvzw-1'))
    
    soup2 = BeautifulSoup(lyrics_div, 'html.parser')
    for br in soup2.find_all("br"):
        br.replace_with("\n")
        
    text = soup2.get_text(separator="\n")
    text = re.sub(r'\n+', '\n', text).strip()
    return text
```

```python
lyrics_goodAsHell = get_clean_lyrics("https://genius.com/Lizzo-good-as-hell-lyrics")
```

```python
print(lyrics_goodAsHell)
```

```python
def extract_urls(url):
    response = requests.get(url)    

    soup1 = BeautifulSoup(response.text)

    track_list = str(soup1.find("ol", class_="AlbumTracklist__Container-sc-123giuo-0 kGJQLs"))
    
    # Parse the HTML content
    soup2 = BeautifulSoup(track_list, 'html.parser')

    # Find all 'a' tags and extract the 'href' attribute
    urls = [a['href'] for a in soup2.find_all('a', href=True)]

    return urls
```

```python
urls = extract_urls("https://genius.com/Lizzo-good-as-hell-lyrics")
```

```python
lyrics_for_album = []
for url in urls:
    lyrics_for_album.append(get_clean_lyrics(url))
```

```python
root = "https://maps.spirithalloween.com"
endpoint = "/api/getAsyncLocations"

headers = {"From":"wfl9zy@virginia.edu", "User-Agent":get_useragent()}
parameters={"template":"domain","level":"domain", 'lat': 40.380028,'lng': -97.910156, "radius":1800, "limit":2000}
```

```python
halloween_mapdata = requests.get(root+endpoint, headers=headers, params=parameters)
```

```python
halloween_stores = halloween_mapdata.json()['markers']
```

```python
info_list = []
for i in range(len(halloween_stores)):
    info_list.append(halloween_stores[i]['info'])
```

```python
def get_all_former_locations(info_list):
    former_locations = []
    for i in range(len(info_list)):
        soup = BeautifulSoup(info_list[i])
        former_locations.append(soup.find("div", class_="address-two").get_text(strip=True))
    return former_locations
```

```python
all_former_locations = get_all_former_locations(info_list)
```

```python
series = pd.Series(all_former_locations)
```

```python
series.value_counts()
```
